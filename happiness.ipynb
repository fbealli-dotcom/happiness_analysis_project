```python
import pandas as pd

# caricamento file
y2015 = pd.read_csv('2015.csv')
y2016 = pd.read_csv('2016.csv')
y2017 = pd.read_csv('2017.csv')
y2018 = pd.read_csv('2018.csv')
y2019 = pd.read_csv('2019.csv')

# creazione della colonna anno per averla poi nel dataset unito
y2015['Year'] = 2015
y2016['Year'] = 2016
y2017['Year'] = 2017
y2018['Year'] = 2018
y2019['Year'] = 2019


```

```python
# prima di concatenare standardizzo e seleziono le colonne che mi interessano
# Mappa per 2015 e 2016
map_15_16 = {
    'Country': 'Country',
    'Happiness Score': 'Score',
    'Economy (GDP per Capita)': 'GDP',
    'Family': 'SocialSupport',
    'Health (Life Expectancy)': 'Health',
    'Freedom': 'Freedom',
    'Trust (Government Corruption)': 'Corruption',
    'Generosity': 'Generosity'
}

# Mappa per 2017
map_17 = {
    'Country': 'Country',
    'Happiness.Score': 'Score',
    'Economy..GDP.per.Capita.': 'GDP',
    'Family': 'SocialSupport',
    'Health..Life.Expectancy.': 'Health',
    'Freedom': 'Freedom',
    'Trust..Government.Corruption.': 'Corruption',
    'Generosity': 'Generosity'
}

# Mappa per 2018 e 2019
map_18_19 = {
    'Country or region': 'Country',
    'Score': 'Score',
    'GDP per capita': 'GDP',
    'Social support': 'SocialSupport',
    'Healthy life expectancy': 'Health',
    'Freedom to make life choices': 'Freedom',
    'Perceptions of corruption': 'Corruption',
    'Generosity': 'Generosity'
}
# Lista delle colonne da mantenere nel dataset finale
columns_to_keep = ['Country', 'Year', 'Score', 'GDP', 'SocialSupport', 'Health', 'Freedom', 'Generosity', 'Corruption']

# Trasformazione dei singoli dataframe
df15 = y2015.rename(columns=map_15_16)[columns_to_keep]
df16 = y2016.rename(columns=map_15_16)[columns_to_keep]
df17 = y2017.rename(columns=map_17)[columns_to_keep]
df18 = y2018.rename(columns=map_18_19)[columns_to_keep]
df19 = y2019.rename(columns=map_18_19)[columns_to_keep]

# Unione finale (Concatenazione verticale)
df_unit = pd.concat([df15, df16, df17, df18, df19], ignore_index=True)

# Recupero della colonna "Region" (presente solo in 2015/2016)
# Creiamo un "anagrafica" dei paesi con la loro regione
regions = pd.concat([y2015[['Country', 'Region']], y2016[['Country', 'Region']]]).drop_duplicates(subset='Country')

# Uniamo la regione al dataset finale tramite un Left Join
df_unit = df_unit.merge(regions, on='Country', how='left')

df_unit.head()
```

```python
import pandas as pd

# 1. Scarichiamo i dati della popolazione mondiale (2015-2019)
# Usiamo un dataset pubblico di Gapminder o della Banca Mondiale
pop_url = "https://raw.githubusercontent.com/datasets/population/master/data/population.csv"
df_pop = pd.read_csv(pop_url)

# 2. Filtriamo solo gli anni 2015-2019
df_pop = df_pop[(df_pop['Year'] >= 2015) & (df_pop['Year'] <= 2019)]

# 3. Rinominiamo 'Country Name' in 'Country' per il merge
df_pop = df_pop.rename(columns={'Country Name': 'Country', 'Value': 'Population'})

# 4. Uniamo al  df_final (assicurati che df_final sia quello creato prima)

df_final = pd.merge(df_unit, df_pop[['Country', 'Year', 'Population']], on=['Country', 'Year'], how='inner')

df_final
```

```python
# Filtriamo solo il 2019, per flourish
df_2019 = df_final[df_final['Year'] == 2019]

# Scarichiamo il file sul tuo PC
df_2019.to_csv('happiness_2019_for_bubbleflourish.csv', index=False)
```

```python
# provo a creare un grafico più statistico-analitico via codice, usando la libreria consigliata dal prof, plotly
import plotly.express as px


df_2019 = df_final[df_final['Year'] == 2019].copy()

# 2. Creazione Grafico
fig = px.scatter(df_2019,
                 x="GDP",
                 y="Score",
                 hover_name="Country",
                 template="plotly_white",
                 opacity=0.6,
                 width=600,
                 height=500,
                 title="<b>Economic Wealth vs. Global Happiness</b>",

                 trendline="lowess",
                 trendline_options=dict(frac=0.6),
                 trendline_color_override="black")

# 3. Aggiungiamo le unità di misura agli assi
fig.update_layout(
    xaxis_title="GDP per Capita (Normalized Index, 0-2)",
    yaxis_title="Happiness Score (Cantril Ladder, 0-10)",
    font_family="Lekton",
    title_font_size=22
)

fig.update_traces(marker=dict(size=6, color='rgba(0, 128, 128, 0.5)',
                             line=dict(width=1, color='DarkSlateGrey')))


fig.show()
```
